# Concaténation des quatre fichiers en un dataframe Pandas

On utilise les dataframes issus de l'examen des fichiers dans la partie précédente. C'est pour cela qu'on n'a pas recours à une boucle `for`.

```python
df = pd.concat((donnees1, donnees2, donnees3, donnees4))
# vérification de la longueur du dataframe concaténé
len(df) == len(donnees1) + len(donnees2) + len(donnees3) + len(donnees4)
```

```python
df.sample(15)
```

```python
df.info()
```

On va prendre chaque colonne, la mettre en forme le cas échéant, vérifier sa conformité.

## Colonne `jour`
On convertit cette colonne en type `date`, puis on vérifie que chaque jour de l'année 2025 est bien présent (comparaison au set `annee`de l'ensemble des jours de 2025).

```python
df["jour"] = pd.to_datetime(df["jour"])
annee = {datetime64("2025-01-01") + timedelta64(i) for i in range(365)}
print(set(df["jour"]) == annee)    # pd.unique(df["jour"]) peut calculer plus vite
                                    # ou annee.difference(set(df["jour"]))
```

## Colone `code_stif_trns`

Il s'agit du code du transporteur. Seule la RATP, code 100 nous intéresse.

```python
pd.unique(df['code_stif_trns'])
```

```python
df = df[df['code_stif_trns'] == 100]
```

## Colonne `code_stif_res`

Il s'agit du code du réseau. Seul le code 110, correspondant au métro, nous intéresse.

```python
pd.unique(df['code_stif_res'])
```

On doit éliminer les données du réseau 'ND'. Cependant on regarde avant la proportion de `nb_vald` (la variable quantitative que l'on va étudier) que cela représente.

```python
df[df['code_stif_res']=='ND']['nb_vald'].sum()/df['nb_vald'].sum()
```

Ce nombre est négligeable.

```python
df = df[df['code_stif_res']=='110']
```

## Colonne `categorie_titre`

```python
pd.unique(df['categorie_titre'])
```

> **On remarque la présente de 'Contrat Solidarité Transport' et 'Contrat Solidarite Transport' ; on corrige de dernier libellé :**

```python
df = df.replace("Contrat Solidarite Transport", "Contrat Solidarité Transport")
pd.unique(df['categorie_titre'])
```

On calcule la proportion de `nb_valid` avec `categorie_titre` 'NON DEFINI' :

```python
df[df['categorie_titre']=='NON DEFINI']['nb_vald'].sum()/df['nb_vald'].sum()
```

## Colonne `nb_vald`

Il s'agit de la seule colonne quantitative que nous allons étudier

```python
df['nb_vald'].groupby(df['categorie_titre']).describe()
```

## Création du fichier source

### Suppression des colonnes inutiles

```python
df = df.drop(['code_stif_trns', 'code_stif_res', 'code_stif_arret', 'ida', 'id_zdc'], axis=1)
df
```

### Agrégation des lignes

```python
# as_index=False : très important pour garder un exportable proprement !
df = df.groupby(by=['jour', 'libelle_arret', 'categorie_titre'], as_index=False).aggregate("sum")
df
```

```python
sorted(pd.unique(df['libelle_arret']))
```

En comparant à lal liste des stations de métro <https://fr.wikipedia.org/wiki/Liste_des_stations_du_m%C3%A9tro_de_Paris>, on note la présence de deux stations supplémentaires dans notre jeu de données :
- ANTONY : pourquoi ?
- FUNICULAIRE : on conserve cette station.

```python
df[df['libelle_arret'] == 'ANTONY']
```

Au vu des très faibles effectifs, il s'agit vraisemblablement d'une erreur. On supprime ces occurrences :

```python
df = df[df['libelle_arret'] != "ANTONY"]
```

On note également 4 stations manquantes :
- *Villejuif-Gustave-Roussy* : ouverte en janvier 2025 et non exploitée par la RATP,
- *Aéroport d'Orly* : tarification spécifique et non exploitée par la RATP,
- *Saint-Denis-Pleyel* : non exploitée par la RATP,
- *Cluny-La-Sorbonne* : *peut-être* du fait de sa liaison avec *Saint-Michel* ? Pas d'explicaton formelle.
