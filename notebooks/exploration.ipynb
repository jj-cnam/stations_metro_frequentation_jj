## Création du fichier source

### Suppression des colonnes inutiles

```python
df = df.drop(['code_stif_trns', 'code_stif_res', 'code_stif_arret', 'ida', 'id_zdc'], axis=1)
df
```

### Recherche de lignes doublonnées

Le jeu données ne devrait *théoriquement* plus comporter de **doublons (même jour, même arrêt, même catégorie de titre)**, mais il faut le verifier :

```python
df_agr = df.groupby(
    ["jour", "libelle_arret", "categorie_titre"]
)["nb_vald"].count()

doublons = df_agr[df_agr > 1]
doublons
```

On trouve un **grand nombre de doublons**, qu'il va falloir examminer :

```python
doublons.sample(10)
```

```python
pd.unique(doublons.reset_index()["libelle_arret"])
```

Ces doublons ne concernent donc que **trois stations**.

```python
len(pd.unique(doublons.reset_index()["jour"]))
```

Mais ils s'étalent sur **toute l'année**.
On examine un échantillon de ces doublons, en faisant tourner plusieurs fois le code suivant :

```python
# Échantillon du jeu de données sur 1 date aléatoire, pour les 3 stations présentant des doublons
df[(df["libelle_arret"].isin(pd.unique(doublons.reset_index()["libelle_arret"]))) \
    & (df["jour"].isin(df["jour"].sample(1)))] \
    .sort_values(["jour", "libelle_arret", "categorie_titre"])
```

**Conclusion : les lignes en doublon comportent des nombres de validations à la fois significatifs *et* différents. En l'absence d'explication, on décide de les *agréger par somme*.**

### Traitement par agrégation des lignes en doublon

```python
# as_index=False : très important pour garder un jeu de données exportable proprement !
df = df.groupby(by=['jour', 'libelle_arret', 'categorie_titre'], as_index=False).aggregate("sum")
df
```
